# Data Cleaning and Preparation

In order to analyze Food Delivery Order data properly, produce accurate results, and turn them into actionable insights, we need to perform data cleaning as our first step. Data cleaning is an essential step in data analysis, since some companies lose millions of dollars due to poor data quality. Therefore, the data cleaning process ensures the following:
1. Values are consistent across the Food Delivery Orders dataset.
2. The Food Delivery Orders dataset does not include duplicate records.
3. Missing values in the Food Delivery Orders dataset are properly handled.
4. Data types in the Food Delivery Orders dataset are standardized.
5. Structural consistency is maintained across the Food Delivery Orders dataset.
6. Outliers in the Food Delivery Orders dataset are identified and handled.
7. Data accuracy is validated before analyzing the Food Delivery Orders dataset.

## 1. Values Consistency
- Standardize 'Distance'(numeric, no km ending)
- Standardize 'Discount construct'(split into columns)
- Standardize 'Instructions'(remove Order:)
- Standardize 'Review'(remove trailing spaces)


```python
df['Distance'].unique()
```

```python
df['Distance under 1 km'] = df['Distance'] == '<1km'
df['Distance'] = df['Distance'].replace('<1km', np.nan)
df['Distance'] = df['Distance'].str.replace('km', '')
df['Distance'] = pd.to_numeric(df['Distance'])
```

```python
df['Distance'].unique()
```

```python
df['Distance under 1 km'].value_counts()
```

```python
for v in df['Discount construct'].unique():
    print(v)
```

```python
# 40% off upto Rs.80
percent_upto = df['Discount construct'].str.contains('% off upto Rs.', na=False)

df.loc[percent_upto, 'Discount type'] = 'percent'

df.loc[percent_upto, 'Discount percent'] = (df.loc[percent_upto, 'Discount construct'].str.split('%').str[0])

df.loc[percent_upto, 'Discount cap'] = (df.loc[percent_upto, 'Discount construct'].str.split('Rs.').str[1])

# Flat Rs.175 off
flat_rs = df['Discount construct'].str.contains('Flat Rs.', na=False) 

df.loc[flat_rs, 'Discount type' ] = 'flat'
df.loc[flat_rs, 'Discount amount'] = (df.loc[flat_rs, 'Discount construct'].str.split('Rs.').str[1]).str.replace(' off', '')


# Flat 15% off
flat_percent = (df['Discount construct'].str.contains('Flat', na=False)
& df['Discount construct'].str.contains('%', na=False))

df.loc[flat_percent, 'Discount type'] = 'percent'
df.loc[flat_percent, 'Discount percent'] = (df.loc[flat_percent, 'Discount construct'].str.split('Flat ').str[1].str.replace('% off', ''))


# Buy 1 Get 1 - ₹1698.00
buy_get = df['Discount construct'].str.contains('Buy ', na=False)

df.loc[buy_get, 'Discount type'] = 'bogo'
df.loc[buy_get, 'Buy quantity'] = (df.loc[buy_get, 'Discount construct'].str.split(' Get').str[0].str.replace('Buy ', ''))
df.loc[buy_get, 'Get quantity'] = (df.loc[buy_get, 'Discount construct'].str.split(' Get ').str[1].str.split(' -').str[0])


# ₹ 0.00 - no discount
no_discount = df['Discount construct'].str.startswith('₹', na=False) 
df.loc[no_discount, 'Discount type'] = 'none'

# Convert to numeric
df['Discount percent'] = pd.to_numeric(df['Discount percent'])
df['Discount cap'] = pd.to_numeric(df['Discount cap'])
df['Discount amount'] = pd.to_numeric(df['Discount amount'])
df['Buy quantity'] = pd.to_numeric(df['Buy quantity'])
df['Get quantity'] = pd.to_numeric(df['Get quantity'])
```

## 3. Handling null values
Decide where missing value is ok and where it is not ok.
- OK (event did not happen): Instructions, Review, Rating, Discount construct, Discount type, Customer complaint tag, Cancellation / Rejection reason, compensation, penalty. Distance null for <1 km (already flagged). KPT / Rider wait when Rejected or Timed out
- Not OK: KPT duration (minutes) and Rider wait time (minutes) when Delivered (197 KPT, 49 rider wait). Leave as NaN (do not fill 0 or mean). Flag those rows. Exclude them from average prep / wait time
- Fill text for Power BI: Instructions = No instructions, Review = No review, Discount construct / Discount type = No discount, Customer complaint tag = No complaint, Cancellation / Rejection reason = Not cancelled or rejected
- Fill money with 0: Restaurant compensation (Cancellation), Restaurant penalty (Rejection)
- Leave as NaN: Rating, Distance, Discount percent, Discount cap, Discount amount, Buy quantity, Get quantity, KPT, Rider wait


```python
df.isna().sum()
```

```python
df.loc[df['KPT duration (minutes)'].isna(), 'Order Status'].value_counts()
```

```python
df.loc[df['Rider wait time (minutes)'].isna(), 'Order Status'].value_counts()
```

```python
for v in df['KPT duration (minutes)'].unique():
    print(v)
```

```python
for v in df['Rider wait time (minutes)'].unique():
    print(v)
```

```python
df['KPT missing on Delivered'] = (df['KPT duration (minutes)'].isna() & (df['Order Status'] == 'Delivered'))

df['Rider wait time missing on Delivered'] = (df['Rider wait time (minutes)'].isna() & (df['Order Status'] == 'Delivered'))
```

```python
df['KPT missing on Delivered'].sum()
```

```python
df['Rider wait time missing on Delivered'].sum()
```

```python
# For PowerBI correct charts
df['Instructions'] = df['Instructions'].fillna('No instructions')
df['Review'] = df['Review'].fillna('No review')
df['Discount construct'] = df['Discount construct'].fillna('No discount')
df['Customer complaint tag'] = df['Customer complaint tag'].fillna('No complaint')
df['Discount type'] = df['Discount type'].fillna('No discount')
df['Cancellation / Rejection reason'] = df['Cancellation / Rejection reason'].fillna('Not cancelled or rejected')
```

```python
df['Restaurant compensation (Cancellation)'] = df['Restaurant compensation (Cancellation)'].fillna(0)

df['Restaurant penalty (Rejection)'] = df['Restaurant penalty (Rejection)'].fillna(0)
```

## 5. Structural Consistency
- rename every column to snake_case (SQL style)
- add a few extra columns for further analysis (item_count, order_date, order_hour, order_weekday)

```python
df = df.rename(columns={
    'Restaurant ID': 'restaurant_id',
    'Restaurant name': 'restaurant_name',
    'Subzone': 'subzone',
    'City': 'city',
    'Order ID': 'order_id',
    'Order Placed At': 'order_placed_at',
    'Order Status': 'order_status',
    'Delivery': 'delivery',
    'Distance': 'distance',
    'Distance under 1 km': 'distance_under_1km',
    'Items in order': 'items_in_order',
    'Instructions': 'instructions',
    'Discount construct': 'discount_construct',
    'Discount type': 'discount_type',
    'Discount percent': 'discount_percent',
    'Discount cap': 'discount_cap',
    'Discount amount': 'discount_amount',
    'Buy quantity': 'buy_quantity',
    'Get quantity': 'get_quantity',
    'Bill subtotal': 'bill_subtotal',
    'Packaging charges': 'packaging_charges',
    'Restaurant discount (Promo)': 'restaurant_discount_promo',
    'Restaurant discount (Flat offs, Freebies & others)': 'restaurant_discount_other',
    'Gold discount': 'gold_discount',
    'Brand pack discount': 'brand_pack_discount',
    'Total': 'total',
    'Rating': 'rating',
    'Review': 'review',
    'Cancellation / Rejection reason': 'cancellation_reason',
    'Restaurant compensation (Cancellation)': 'restaurant_compensation',
    'Restaurant penalty (Rejection)': 'restaurant_penalty',
    'KPT duration (minutes)': 'kpt_minutes',
    'Rider wait time (minutes)': 'rider_wait_minutes',
    'KPT missing on Delivered': 'kpt_missing_on_delivered',
    'Rider wait time missing on Delivered': 'rider_wait_missing_on_delivered',
    'Order Ready Marked': 'order_ready_marked',
    'Customer complaint tag': 'customer_complaint_tag',
    'Customer ID': 'customer_id',
})
```

```python
df.columns
```

```python
df['item_count'] = (
    df['items_in_order']
    .str.findall(r'(\d+) x ')
    .apply(lambda nums: sum(int(n) for n in nums))
)

df['order_date'] = df['order_placed_at'].dt.date

df['order_hour'] = df['order_placed_at'].dt.hour

df['order_weekday'] = df['order_placed_at'].dt.day_name()
```

## 6. Outliers are identified and handled
- Checked total, item_count, kpt_minutes, rider_wait_minutes, distance, rating
- Big bills and long prep/wait times are real (party orders, slow kitchen). Nothing dropped
- One delivered order has kpt_minutes = 0 and ready marked Missed. That is not a real prep time, so set it to NaN


```python
df[['total', 'item_count', 'kpt_minutes', 'rider_wait_minutes', 'distance', 'rating']].describe()
```

```python
# Unrealistic kpt_minutes = 0 when delivered (fix)
df.loc[(df['kpt_minutes'] == 0) & (df['order_status'] == 'Delivered'), 'kpt_minutes'] = np.nan
```
